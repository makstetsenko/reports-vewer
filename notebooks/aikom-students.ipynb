```python
import re

from rapidfuzz import fuzz

def search_by(column_name, values: list[str], threshold: int = 80):
    if not values:
        return pd.Series(False, index=df.index)

    def matches(surname):
        if pd.isna(surname):
            return False

        return any(
            fuzz.partial_ratio(value.casefold(), str(surname).casefold()) >= threshold
            for value in values
            if value
        )

    return df[column_name].apply(matches)
```
